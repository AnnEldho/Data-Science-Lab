## **Histogram**

```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()

filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)

features = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

df[features].hist(
    bins=10,
    figsize=(10, 8),
    edgecolor="black"
)

plt.suptitle("Distribution of Iris Features")
plt.tight_layout()
plt.show()
```

### **Quartile (Box) Plot**

```python
import matplotlib.pyplot as plt
import seaborn as sns

features = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, feature in zip(axes.flatten(), features):
    sns.boxplot(
        data=df,
        x="Species",
        y=feature,
        ax=ax
    )
    ax.set_title(feature)

plt.suptitle("Box Plots of Iris Features by Species")
plt.tight_layout()
plt.show()
```

### **Distribution Chart**

```python
import seaborn as sns
import matplotlib.pyplot as plt

features = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, feature in zip(axes.flatten(), features):
    sns.histplot(
        data=df,
        x=feature,
        hue="Species",
        kde=True,
        bins=10,
        ax=ax
    )
    ax.set_title("Distribution of " + feature)

plt.suptitle("Distribution of Iris Features by Species")
plt.tight_layout()
plt.show()
```

## **Scatterplot**

```python
import seaborn as sns
import matplotlib.pyplot as plt

features = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

pairs = [
    ("SepalLengthCm", "SepalWidthCm"),
    ("SepalLengthCm", "PetalLengthCm"),
    ("SepalLengthCm", "PetalWidthCm"),
    ("PetalLengthCm", "PetalWidthCm")
]

for ax, (x, y) in zip(axes.flatten(), pairs):
    sns.scatterplot(
        data=df,
        x=x,
        y=y,
        hue="Species",
        ax=ax,
        s=60
    )
    ax.set_title(f"{x} vs {y}")

plt.suptitle("Scatterplots of Iris Features")
plt.tight_layout()
plt.show()
```

### **Scatter Matrix**

```python
import seaborn as sns
import matplotlib.pyplot as plt

features = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

sns.pairplot(
    df[features + ["Species"]],
    hue="Species",
    diag_kind="hist"
)

plt.suptitle("Scatter Matrix of Iris Dataset", y=1.02)
plt.show()
```

### **Bubble Chart**

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))

bubble_size = df["PetalLengthCm"] * df["PetalWidthCm"] * 20

sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="SepalWidthCm",
    size=bubble_size,
    hue="Species",
    sizes=(30, 500),
    alpha=0.6
)

plt.title("Bubble Chart of Iris Features")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")

plt.show()
```

### **Density Chart**

```python
import seaborn as sns
import matplotlib.pyplot as plt

features = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))

for ax, feature in zip(axes.flatten(), features):
    sns.kdeplot(
        data=df,
        x=feature,
        hue="Species",
        fill=True,
        ax=ax
    )
    ax.set_title("Density of " + feature)

plt.suptitle("Density Distribution of Iris Features")
plt.tight_layout()
plt.show()
```

### **Parallel Chart**

```python
import pandas as pd
import matplotlib.pyplot as plt
from pandas.plotting import parallel_coordinates

# Select the required columns
parallel_data = df[
    [
        "Species",
        "SepalLengthCm",
        "SepalWidthCm",
        "PetalLengthCm",
        "PetalWidthCm"
    ]
]

# Create parallel coordinates chart
plt.figure(figsize=(12, 6))

parallel_coordinates(
    parallel_data,
    "Species"
)

plt.title("Parallel Coordinates Chart of Iris Dataset")
plt.xlabel("Features")
plt.ylabel("Values")

plt.xticks(rotation=20)
plt.grid(True)

plt.show()
```

### **Deviation Chart**

```python
import pandas as pd
import matplotlib.pyplot as plt

features = [
    "SepalLengthCm",
    "SepalWidthCm",
    "PetalLengthCm",
    "PetalWidthCm"
]

mean_values = df[features].mean()

deviation = df[features] - mean_values

plt.figure(figsize=(12, 6))

plt.plot(
    deviation.index,
    deviation["SepalLengthCm"],
    label="Sepal Length"
)

plt.plot(
    deviation.index,
    deviation["SepalWidthCm"],
    label="Sepal Width"
)

plt.plot(
    deviation.index,
    deviation["PetalLengthCm"],
    label="Petal Length"
)

plt.plot(
    deviation.index,
    deviation["PetalWidthCm"],
    label="Petal Width"
)

plt.axhline(
    y=0,
    linestyle="--"
)

plt.title("Deviation of Iris Features from Their Mean")
plt.xlabel("Observation")
plt.ylabel("Deviation from Mean")
plt.legend()

plt.show()
```

### **Andrews Curves**

```python
import matplotlib.pyplot as plt
from pandas.plotting import andrews_curves

# Select the required columns
andrews_data = df[
    [
        "Species",
        "SepalLengthCm",
        "SepalWidthCm",
        "PetalLengthCm",
        "PetalWidthCm"
    ]
]

# Create Andrews Curves
plt.figure(figsize=(12, 6))

andrews_curves(
    andrews_data,
    "Species"
)

plt.title("Andrews Curves of Iris Dataset")
plt.xlabel("Curve Position")
plt.ylabel("Feature Value")

plt.grid(True)
plt.show()
```
