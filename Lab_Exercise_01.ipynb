### **2.Pandas**

```python
import pandas as pd

data = {
    "Name":["Ann","John","Sara"],
    "Age":[22,21,23],
    "Marks":[95,85,90]
}

df = pd.DataFrame(data)

print("Original DataFrame")
print(df)

print("\nShape")
print(df.shape)

print("\nColumns")
print(df.columns)

print("\nFirst Rows")
print(df.head(1))

print("\nLast Two Rows")
print(df.tail(2))

print("\nInformation")
df.info()

print("\nDescription")
print(df.describe())

print("\nAverage Marks")
print(df["Marks"].mean())

print("\nHighest Marks")
print(df["Marks"].max())

print("\nLowest Marks")
print(df["Marks"].min())

print("\nTotal Marks")
print(df["Marks"].sum())

print("\nSorted by Marks")
print(df.sort_values("Marks"))

print("\nUsing loc")
print(df.loc[0])

print("\nUsing iloc")
print(df.iloc[2])

print("\nAfter Dropping Age Column")
print(df.drop(columns=["Age"]))
```

### **Matplotlib**

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,20,15,25,30]

# Line Graph
plt.plot(x, y, label="Sales")
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Line Graph")
plt.legend()
plt.grid(True)
plt.xlim(1,5)
plt.ylim(0,35)
plt.show()

# Bar Graph
plt.bar(x, y)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Bar Graph")
plt.show()

# Scatter Plot
plt.scatter(x, y)
plt.xlabel("Days")
plt.ylabel("Sales")
plt.title("Scatter Plot")
plt.show()

# Histogram

marks = [55,60,62,65,67,68,70,72,73,75,
         76,78,80,81,82,84,85,87,88,90,
         91,92,93,95]

plt.hist(marks, bins=8, edgecolor='black')

plt.title("Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")

plt.show()
# Pie Chart
subjects = ["Python","Java","C","AI"]
students = [40,30,20,10]

plt.pie(students,
        labels=subjects,
        autopct="%1.1f%%")

plt.title("Students Preference")
plt.show()
```

### **Seaborn**

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Creating Dataset
data = {
    "Name": ["Ann", "John", "Sara", "Tom", "David"],
    "Age": [20, 21, 22, 23, 24],
    "Marks": [85, 90, 78, 92, 88],
    "Department": ["MCA", "MCA", "MBA", "MBA", "MCA"]
}

df = pd.DataFrame(data)

# Line Plot
sns.lineplot(data=df, x="Age", y="Marks")
plt.title("Line Plot")
plt.show()

# Scatter Plot
sns.scatterplot(data=df, x="Age", y="Marks")
plt.title("Scatter Plot")
plt.show()


# Bar Plot
sns.barplot(data=df, x="Name", y="Marks")
plt.title("Bar Plot")
plt.show()

# Histogram
sns.histplot(data=df, x="Marks", bins=5)
plt.title("Histogram")
plt.show()

# Box Plot
sns.boxplot(data=df, x="Department", y="Marks")
plt.title("Box Plot")
plt.show()

# Count Plot
sns.countplot(data=df, x="Department")
plt.title("Count Plot")
plt.show()

# Heatmap
corr = df[["Age", "Marks"]].corr()

sns.heatmap(corr, annot=True, cmap="Blues")
plt.title("Heatmap")
plt.show()

# Pair Plot
sns.pairplot(df[["Age", "Marks"]])
plt.show()
```

### **Bokeh**

```python
from bokeh.plotting import figure, show
from bokeh.io import output_notebook

# Display plots inside Colab
output_notebook()

days = [1, 2, 3, 4, 5]
sales = [120, 180, 150, 200, 170]

p = figure(
    title="Sales Report",
    x_axis_label="Days",
    y_axis_label="Sales",
    width=600,
    height=400
)

p.line(days, sales, line_width=3, color="blue", legend_label="Sales")

p.scatter(
    x=days,
    y=sales,
    marker="circle",
    size=10,
    color="red",
    legend_label="Points"
)

p.vbar(
    x=days,
    top=sales,
    width=0.4,
    color="green",
    alpha=0.5,
    legend_label="Bar"
)

show(p)
```

### **Scikit Learn**

```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier

# Load dataset
iris = load_iris()

# Create model
model = DecisionTreeClassifier()

# Train model
model.fit(iris.data, iris.target)

# Predict
prediction = model.predict([[5.1, 3.5, 1.4, 0.2]])

print("Predicted Class:", prediction)
print("Flower Name:", iris.target_names[prediction][0])
```

### **TensorFlow**

```python
import tensorflow as tf

# Constant Tensor
a = tf.constant([10, 20, 30])

# Variable Tensor
b = tf.Variable([1, 2, 3])

print("Constant Tensor:", a)
print("Variable Tensor:", b)

# Addition
print("Addition:", tf.add(a, b))

# Multiplication
print("Multiplication:", tf.multiply(a, b))

# Matrix Multiplication
m1 = tf.constant([[1, 2], [3, 4]])
m2 = tf.constant([[5, 6], [7, 8]])

print("Matrix Multiplication:")
print(tf.matmul(m1, m2))

# Sum
print("Sum:", tf.reduce_sum(a))

# Mean
print("Mean:", tf.reduce_mean(tf.cast(a, tf.float32)))
```

### **SciPy**

```python
from scipy import stats
from scipy.special import factorial
from scipy.integrate import quad
import numpy as np

data = [10, 20, 20, 30, 40]

print("Square Root of 25 =", np.sqrt(25))

print("Factorial of 5 =", factorial(5))

print("Mean =", stats.tmean(data))

print("Median =", np.median(data))

print("Mode =", stats.mode(data, keepdims=True))

def f(x):
    return x**2

result, error = quad(f, 0, 2)

print("Integration =", result)
```

### **Keras**

```python
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

X = np.array([[1],[2],[3],[4],[5]])
y = np.array([2,4,6,8,10])

model = Sequential()

model.add(Dense(1, input_dim=1))

model.compile(optimizer="adam", loss="mse")

model.fit(X, y, epochs=100, verbose=0)

model.summary()

result = model.predict(np.array([[6]]), verbose=0)

print("Prediction:", result)
```

### **NLTK**

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')

from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk import pos_tag, FreqDist

text = "Python is a programming language. It is easy to learn."

print("Sentences:")
print(sent_tokenize(text))

words = word_tokenize(text)
print("\nWords:")
print(words)

stop_words = stopwords.words("english")
filtered = [word for word in words if word.lower() not in stop_words]

print("\nAfter Removing Stop Words:")
print(filtered)

print("\nPOS Tagging:")
print(pos_tag(filtered))

print("\nWord Frequency:")
freq = FreqDist(filtered)
print(freq)
```

### **Beautiful** **Soup**

```python
from bs4 import BeautifulSoup


html = """
<html>
<head>
<title>Student Details</title>
</head>
<body>

<h1>Welcome to BeautifulSoup</h1>

<p>Name : Ann Mary</p>
<p>Course : MCA</p>

<a href="https://www.python.org">Python</a>

</body>
</html>
"""

soup = BeautifulSoup(html, "html.parser")

print("Title:")
print(soup.title.get_text())

print("\nHeading:")
print(soup.find("h1").get_text())

print("\nParagraphs:")
for p in soup.find_all("p"):
    print(p.get_text())

print("\nLink:")
print(soup.find("a")["href"])

print("\nFormatted HTML:")
print(soup.prettify())
```

### **Plotly**

```python
import plotly.express as px
import pandas as pd

data = {
    "Student": ["Ann", "John", "Sara", "Tom", "Eva"],
    "Marks": [85, 90, 78, 88, 95],
    "Age": [21, 22, 20, 23, 21]
}

df = pd.DataFrame(data)

fig = px.line(df,
              x="Student",
              y="Marks",
              title="Line Chart")
fig.show()

fig = px.bar(df,
             x="Student",
             y="Marks",
             color="Student",
             title="Bar Chart")
fig.show()

fig = px.scatter(df,
                 x="Age",
                 y="Marks",
                 color="Student",
                 size="Marks",
                 title="Scatter Plot")
fig.show()

fig = px.pie(df,
             names="Student",
             values="Marks",
             title="Pie Chart")
fig.show()

fig = px.histogram(df,
                   x="Marks",
                   nbins=5,
                   title="Histogram")
fig.show()

fig = px.box(df,
             y="Marks",
             title="Box Plot")
fig.show()
```
